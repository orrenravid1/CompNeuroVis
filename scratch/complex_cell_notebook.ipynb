```python
import os
#os.environ['CNV_NOTEBOOK_RFB'] = '1'
os.environ['CNV_NOTEBOOK_RENDER_PROCESS'] = '1'

import pathlib
from neuron import h
import compneurovis as cnv
from compneurovis.backends.neuron.utils import load_swc_neuron

# Robustly resolve res/ from the installed package location (repo root).
repo = pathlib.Path(cnv.__file__).resolve().parents[2]
swc_path = str(repo / 'res' / 'Animal_2_Basal_2.CNG.swc')

sections = load_swc_neuron(swc_path)

for sec in sections:
    sec.insert('hh')
    # Higher nseg on dendrites = finer spatial resolution = heavier sim + render.
    if 'soma' not in sec.name().lower():
        sec.nseg = 1

soma = next(sec for sec in sections if 'soma' in sec.name().lower())
n_seg = sum(sec.nseg for sec in sections)
print(f'sections={len(sections)}  total segments={n_seg}')

stim = h.IClamp(soma(0.5))
stim.delay = 0.0
stim.dur = 1e9
stim.amp = 0.5

h.dt = 0.025
h.celsius = 6.3
h.finitialize(-65.0)

sim = cnv.source(cnv.neuron.attach(sections=sections))
sim.morphology()
def set_clamp_amp(ctx, value):
    stim.amp = float(value)

sim.slider(
    'clamp_amp',
    label='IClamp amplitude (nA)',
    get=lambda: float(stim.amp),
    set=set_clamp_amp,
    min=-0.5,
    max=2.0,
    default=float(stim.amp),
)
```
